4.  Implement Gradient Descent Algorithm to find the local minima of a function. 
For example, find the local minima of the function y=(x+3)² starting from the point x=2. 

```python
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3)**2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2
learning_rate = 0.1
iterations = 20

x_values = []

for i in range(iterations):
    x_values.append(x)

    x = x - learning_rate * derivative(x)

    print("Iteration:", i+1, "X:", x, "Y:", function(x))

print("Minimum value of x:", x)
print("Minimum value of y:", function(x))

print("Derivative at minimum:", derivative(x))
```

```python
x = range(-6, 4)

y = [function(i) for i in x]

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent")
plt.grid()
plt.show()
```
